The Text Formatter & Validator accepts a full name and an email-like string from the user, formats the name using title case, trims extra whitespace, validates whether the input contains only alphabetic characters or is alphanumeric, and displays a formatted greeting using an f-string, using only string methods and operators (no loops).

```python
name = input("Enter your name: ").strip()
email = input("Enter email: ").strip()
formatted_name = name.title()
print(formatted_name.strip())

if formatted_name.replace(" ","").isalpha():
  print("Valid alphabet")
else:
  print("Invalid Name")

if email.replace("@", "").replace(".","").isalnum():
  print("Valid email:")
else:
  print("Invalid email")

print(f"Hello {formatted_name}, Good day!!")
```
