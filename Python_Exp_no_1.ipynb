```python
principle = float(input("Enter the Principle: "))
Time = float(input("Enter the Time in years: "))
Rate = float(input("Enter the Rate: "))
SI = float((principle*Time*Rate)/100)
print(f"Simple Interest:{SI}")
A = float(principle*(1+(Rate/100))**(Time))
CI = float(A-principle)
print(f"Compound Interest:{CI}")
```

```python
#Write a Python program to convert temperature from Celsius to Fahrenheit and vice versa.
print("--- TEMPERATURE CONVERTER ---")
print("1. Convert Celsius to Fahrenheit")
print("2. Convert Fahrenheit to Celsius")

choice = float(input("Enter your choice (1 or 2): "))
match choice:
    case 1:
        celsius = float(input("Enter temperature in Celsius: "))
        fahrenheit = (celsius * (9/5)) + 32
        print(f"Temperature in Fahrenheit:{fahrenheit}")

    case 2:
        fahrenheit = float(input("Enter temperature in Fahrenheit: "))
        celsius = (fahrenheit - 32)*(5/9)
        print(f"Temperature in Celsius:{celsius}")
```

```python
#Write a Python program to demonstrate local and global variables.
print("--- LOCAL AND GLOBAL VARIABLES ---")
global_var = "This is GLOBAL!"
def local():
  local_var = "This is LOCAL!"
  print("Inside the function:")
  print(f"This is local variable:{local_var}")
  print(f"This is global variable:{global_var}")
local()
print("Outside the function:")
print(f"This is global variable:{global_var}")
```

```python
#14.Write a Python program to declare a constant PI_VALUE and use it to calculate the circumference of a circle.
PI_VALUE = 3.14
radius = float(input("Enter the radius of the circle: "))
circumference = 2 * PI_VALUE * radius
print(f"Constant PI_VALUE used:{PI_VALUE}")
print(f"Radius entered:{radius}")
print("Circumference:", circumference)
```
