```python
n = int(input("Enter the number of subjects: "))
Subjects = []
for i in range (n):
  Subjects.append(input("Enter the name of subjects: "))
print(Subjects)
Marks = []
for i in range(n):
  Marks.append(int(input("Enter the marks of the subjects respectively: ")))
print(Subjects, "=", Marks)
print("Total_marks = ", sum(Marks))
print("Highest_Marks = ", max(Marks))
print("Lowest_Marks = ", min(Marks))
Average = "Average_Marks = ", sum(Marks)/n
print(Average)
Marks.sort()
print("Sorted_Marks = ", Marks)
ans = input("Do you want to insert marks (yes/no): ")
if ans == "yes":
  change = input("Do you want to replace marks or add a new marks for new subeject or remove marks of a subject (replace/add/remove): ")
  if change == "replace":
    position = (int(input("Enter the position: "))) - 1
    i = position
    replaced_marks = int(input("Enter the mark to replace: "))
    Marks[i] = replaced_marks
    print("Updated Marks: ", Marks)
    print("New_Total_marks = ", sum(Marks))
    print("New_Highest_Marks = ", max(Marks))
    print("New_Lowest_Marks = ", min(Marks))
    New_average = "New_Average_Marks = ", sum(Marks)/n
    print(New_average)
  elif change == "add":
    Subjects.append(input("Enter the new subject: "))
    Marks.append(int(input("Enter the marks of the new subject: ")))
    n+=1
    print(Subjects, "=", Marks)
    print("Updated Marks: ", Marks)
    print("New_Total_marks = ", sum(Marks))
    print("New_Highest_Marks = ", max(Marks))
    print("New_Lowest_Marks = ", min(Marks))
    New_average = "New_Average_Marks = ", sum(Marks)/n
    print(New_average)
  elif change == "remove":
    position = (int(input("Enter the position: "))) - 1
    mark_to_remove = int(input("Enter the mark to remove: "))
    Marks.remove(mark_to_remove)
    print("Updated Marks: ", Marks)
    print("New_Total_marks = ", sum(Marks))
    print("New_Highest_Marks = ", max(Marks))
    print("New_Lowest_Marks = ", min(Marks))
    New_average = "New_Average_Marks = ", sum(Marks)/n
    print(New_average)
  Marks.sort()
  print("New sorted and Updated Marks: ", Marks)
  Student_details = (input("Enter Student name: "), input("Enter roll number: "), New_average)
  Name, Roll_number, Average = Student_details
  print(Name, Roll_number, Average)
else :
  Student_details = (input("Enter Student name: "), input("Enter roll number: "), Average)
  Name, Roll_number, Average = Student_details
  print(Name, Roll_number, Average)
```
